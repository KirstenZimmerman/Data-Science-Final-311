```python
import numpy as np
import pandas as pd
from sklearn.metrics import accuracy_score, f1_score, roc_auc_score, confusion_matrix
```

```python
train_clean = pd.read_csv("../data/processed/train_clean.csv")
test_clean = pd.read_csv("../data/processed/test_clean.csv")
```

```python
smart_rule_train = (
    (train_clean["Support Calls"] >= 5) | 
    (train_clean["Payment Delay"] >= 20) | 
    (train_clean["Total Spend"] < 500) | 
    (train_clean["Age"] >= 50)
)

train_clean["smart_baseline_pred"] = smart_rule_train.astype(int)
```

```python
y_true = train_clean["Churn"]
y_pred = train_clean["smart_baseline_pred"]
```

```python
accuracy = accuracy_score(y_true, y_pred)
f1 = f1_score(y_true, y_pred)
auc = roc_auc_score(y_true, y_pred)
conf = confusion_matrix(y_true, y_pred)

print(f"Accuracy: {accuracy:.4f}")
print()
print(f"F1-score: {f1:.4f}")
print()
print(f"ROC AUC: {auc:.4f}")
print()
print("Confusion Matrix:")
print(conf)
```

```python
smart_rule_test = (
    (test_clean["Support Calls"] >= 5)
    | (test_clean["Payment Delay"] >= 20)
    | (test_clean["Total Spend"] < 500)
    | (test_clean["Age"] >= 50)
)

test_clean["Churn"] = smart_rule_test.astype(int)
```

```python
submission_smart = pd.DataFrame({
    "CustomerID": test_clean["CustomerID"],
    "Churn": test_clean["Churn"],
})

submission_smart.to_csv( "data/submissions/baseline.csv", index=False)
```
