```python
import pandas as pd

from sklearn.dummy import DummyClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report,
    roc_auc_score,
)
```

```python
train = pd.read_csv("../data/processed/train_clean.csv")
test = pd.read_csv("../data/processed/test_clean.csv")
train.head()
```

```python
baseline_features = [
    "Total Spend",
    "Usage Frequency",
    "Support Calls",
]

X = train[baseline_features]
y = train["Churn"]
```

```python
X_train, X_val, y_train, y_val = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=1234,
    stratify=y
)
```

```python
model = DummyClassifier(strategy="constant", constant=0)
model.fit(X_train, y_train)
```

```python
y_pred_val = model.predict(X_val)
y_proba_val = model.predict_proba(X_val)[:, 1]
```

```python
dumb_model = DummyClassifier(strategy="constant", constant=0)
dumb_model.fit(X_train, y_train)

y_pred = dumb_model.predict(X_val)
```

```python
print("Accuracy:")
print(accuracy_score(y_val, y_pred))
print()
print("Confusion matrix:")
print(confusion_matrix(y_val, y_pred))
print()
print("AUC:")
print(roc_auc_score(y_val, y_proba_val))
# print("Classification report:") #outputs a warning
# print(classification_report(y_val, y_pred))
```

```python
X_test = test[baseline_features]
test_pred = model.predict(X_test)

submission = pd.DataFrame({
    "CustomerID": test["CustomerID"],
    "Churn": test_pred
})

submission.to_csv("../data/submissions/dumb_baseline.csv", index=False)
submission.head()
```
